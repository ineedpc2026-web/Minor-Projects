### Exercise 3 - extend vs append  ·  *medium*

Run both cells. Predict the length of each list **before** you run it.

```python
List = ["Michael", 10.1]
List.extend(["pop", 10])
print(List)          # Prediction - ['Michael', 10.1, 'pop', 10]
print(len(List))     # Prediction - 4
```

Now the other one.

```python
L = ["Michael", 10.1]
L.append(["pop", 10])
print(L)          # ['Michael', 10.1, ['pop', 10]]
print(len(L))     # 3

# extend unpacks. append does not. That is the entire difference,
# and it is responsible for a genuinely silly number of bugs.
```

### Exercise 5 - Aliasing  ·  *hard*

**Predict the output before you run it.** Then run it. Then read the explanation.

```python
A = ["hard rock", 10, 1.2]
B = A
A[0] = "banana"

print("A:", A)        # ['banana', 10, 1.2]
print("B:", B)        # ['banana', 10, 1.2]

# Two labels. One box. If you edit the box, both labels see it.
# For numbers and strings this does not happen. For lists it does.
```

### Exercise 6 - Fix the aliasing  ·  *hard*

Now make `B` an actual, independent copy.

```python
A = ["hard rock", 10, 1.2]
B = A[:]              # a full slice = a clone. A new box.
A[0] = "banana"

print("A:", A)        # ['banana', 10, 1.2]
print("B:", B)        # ['hard rock', 10, 1.2]

# Other ways to do the same thing:
C = A.copy()
D = list(A)
```

### Exercise 11 - sort vs sorted  ·  *hard*

**Predict both outputs before running.**

```python
ratings = (10, 9, 6, 5, 10, 8)

a = sorted(ratings)
print(a)              # (5, 6, 8, 9, 10, 10) because Sorted creates a new list
print(ratings)        # (10, 9, 6, 5, 10, 8) and old one remains untouched

L = [10, 9, 6, 5]
b = L.sort()
print(L)              # [5, 6, 9, 10] as .sort changes the original list
print(b)              # it returns none
```

### Exercise 12 - Put it together  ·  *hard*

A messy string of scores arrived. Clean it into a sorted list of numbers,
and print the top three.

```python
raw = "88, 92, 75, 95, 61, 89"

pieces = raw.split(",")
scores = [int(num.strip()) for num in pieces]
ranked = sorted(scores, reverse=True)         # arrange from highest first
top3 = ranked[:3]                             # first three

print("All scores:", scores)
print("Ranked:    ", ranked)
print("Top three: ", top3)
```
