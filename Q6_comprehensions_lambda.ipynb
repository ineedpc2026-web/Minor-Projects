Two ideas only:
- A **comprehension** writes a loop in one line.
- A **lambda** is a small function with no name.

**Read a comprehension out loud, left to right:**

```python
[n * n for n in nums]
```
*"Give me n times n, for each n in nums."*

Left part = what you want. Right part = where it comes from.

Do them in order. They build.

### Exercise 1 - Your first comprehension  ·  *easy*

Square every number.

````python
nums = [1, 2, 3, 4, 5]

squares = [n * n for n in nums]

print(squares)     # Expect: [1, 4, 9, 16, 25]
````

### Exercise 2 - Do something to every item  ·  *easy*

````python
names = ["asha", "ravi", "meera"]

print([n.upper() for n in names])     # Expect: ['ASHA', 'RAVI', 'MEERA']
print([len(n) for n in names])     # Expect: [4, 4, 5]      (lengths)
print([n[0] for n in names])     # Expect: ['a', 'r', 'm'] (first letters)
````

### Exercise 3 - Add a filter  ·  *easy*

An `if` at the end keeps only some items.

````python
nums = [1, 2, 3, 4, 5, 6]

evens = [n for n in nums if n % 2 == 0]
print(evens)      # Expect: [2, 4, 6]

big = [n for n in nums if n > 3]
print(big)        # Expect: [4, 5, 6]
````

### Exercise 4 - Change AND filter  ·  *medium*

Double only the odd numbers.

````python
nums = [1, 2, 3, 4, 5]

result = [n * 2 for n in nums if n % 2 != 0]

print(result)     # Expect: [2, 6, 10]
````

### Exercise 5 - Comprehension on a string  ·  *medium*

Get the vowels out of a word.

````python
word = "Analytics"

vowels = [c for c in word.lower() if c in "aeiou"]

print(vowels)          # Expect: ['a', 'a', 'i']
print(len(vowels))     # Expect: 3
````

### Exercise 6 - Clean a messy list  ·  *medium*

Strip the spaces and cast to int, in one line.

````python
raw = ["  88 ", "92", " 75  ", "95 "]

scores = [int(s.strip()) for s in raw]

print(scores)          # Expect: [88, 92, 75, 95]
print(sum(scores))     # Expect: 350
````

### Exercise 7 - Dictionary comprehension  ·  *medium*

Curly braces, and `key: value` on the left.

````python
prices = {"tea": 10, "coffee": 25, "juice": 40}

# Raise every price by 10%.
hiked = {k : round(v * 1.1 , 2) for k, v in prices.items()}
print(hiked)     # Expect: {'tea': 11.0, 'coffee': 27.5, 'juice': 44.0}

# Keep only the items under 30.
cheap = {k: v for k, v in prices.items() if v < 30}
print(cheap)     # Expect: {'tea': 10, 'coffee': 25}
````

### Exercise 8 - Set comprehension  ·  *medium*

Same idea. Curly braces, no colon. You get uniqueness for free.

````python
words = ["apple", "banana", "avocado", "blueberry", "apricot"]

first_letters = {w[0] for w in words}

print(first_letters)     # Expect: {'a', 'b'}  - unique automatically
````

### Exercise 9 - Comprehension on nested data  ·  *hard*

This is the shape of real data. Get comfortable here.

````python
students = [
    {"name": "Asha",  "city": "Delhi",  "marks": [88, 92, 79]},
    {"name": "Ravi",  "city": "Mumbai", "marks": [75, 81, 90]},
    {"name": "Meera", "city": "Delhi",  "marks": [95, 89, 84]},
]

print([s["name"] for s in students])                    # Expect: ['Asha', 'Ravi', 'Meera']
print([s["name"] for s in students if s["city"] == "Delhi"])            # Expect: ['Asha', 'Meera']  (Delhi only)
print([sum(s["marks"]) for s in students])                    # Expect: [259, 246, 268]   (totals)
````

### Exercise 10 - Your first lambda  ·  *easy*

A lambda is a small function with no name. The bit after the colon is what it gives back.

````python
double = lambda x: x * 2
print(double(5))         # Expect: 10

add = lambda a, b: a + b
print(add(3, 4))         # Expect: 7

is_adult = lambda age: age > 18
print(is_adult(25))      # Expect: True
print(is_adult(12))      # Expect: False
````

### Exercise 11 - sorted with a key  ·  *hard*

This is the lambda you will actually use. `key` tells `sorted` **what to sort by**.

````python
people = [("Asha", 88), ("Ravi", 75), ("Meera", 95)]

print(sorted(people))                           # sorts by name - the default

print(sorted(people, key=lambda p: p[1]))       # by marks, lowest first
print(sorted(people, key=lambda p: p[1], reverse=True))   # by marks, highest first

prices = {"tea": 10, "coffee": 25, "juice": 40}
print(sorted(prices.items(), key=lambda pair: pair[1], reverse=True))   # dearest first
````

### Exercise 12 - Sort nested data  ·  *hard*

Rank the students by total marks, best first, and print just the names.

````python
students = [
    {"name": "Asha",  "marks": [88, 92, 79]},
    {"name": "Ravi",  "marks": [75, 81, 90]},
    {"name": "Meera", "marks": [95, 89, 84]},
]

ranked = sorted(students, key=lambda s: sum(s["marks"]), reverse=True)
print(ranked)
print([s["name"] for s in ranked])     # Expect: ['Meera', 'Asha', 'Ravi']
````

### Exercise 13 - map and filter  ·  *medium*

Both return a lazy object, so wrap them in `list()` to see anything.

````python
nums = [1, 2, 3, 4]

print(list(map(lambda x: x * 2, nums)))     # Expect: [2, 4, 6, 8]
print(list(filter(lambda x: x > 2, nums)))     # Expect: [3, 4]

# Now write the SAME two things as comprehensions.
print([num * 2 for num in nums])      # Expect: [2, 4, 6, 8]
print([num for num in nums if num > 2])      # Expect: [3, 4]
````

### Exercise 14 - Put it all together  ·  *hard*

A messy string of orders arrived. Produce a clean report.

`"tea:3, coffee:2, juice:5, tea:1"`

Print:
1. a list of the item names, no duplicates
2. a dictionary of `{item: total quantity}`
3. the items sorted by quantity, highest first

````python
raw = "tea:3, coffee:2, juice:5, tea:1"

# Step 1: split on the comma, then strip each piece
pairs = raw.split(', ')
print(pairs)

# Step 2: split each piece on the colon -> [['tea','3'], ['coffee','2'], ...]
split_pairs = [item.split(':') for item in pairs]
print(split_pairs)

# Step 3: unique item names
items = set(item[0] for item in split_pairs)
print("Items:", items)

# Step 4: total quantity per item
totals = {item: sum(int(p[1]) for p in split_pairs if p[0] == item) for item in items}
print("Totals:", totals)

# Step 5: sorted, highest first
ranked = sorted(totals.items() , key=lambda item: item[1], reverse=True)
print("Ranked:", ranked )
````
