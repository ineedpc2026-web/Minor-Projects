### Exercise 8 - Sets kill duplicates  ·  *easy*

Predict output before running cell.

```python
album_list = ["Thriller", "Back in Black", "Thriller", "AC/DC", "AC/DC"]

album_set = set(album_list)

print(album_set)          # {"Thriller", "Back in Black", "AC/DC"} -> Sets are Unordered and have no Duplicate items
print(len(album_list))    # 5
print(len(album_set))     # 3

print(len(album_list) - len(album_set))    # 2 - Shows number of duplicates
```

### Exercise 10 - Set maths  ·  *medium*

Predict output before running cells.

```python
a = {"AC/DC", "Back in Black", "Thriller"}
b = {"AC/DC", "Back in Black", "The Dark Side of the Moon"}

print(a & b)     # {"AC/DC", "Back in Black"}
print(a | b)     # {"AC/DC", "Back in Black", "thriller", "The Dark Side of the Moon"}
print(a - b)     # {"Thriller"}

print(a.intersection(b))         # {"AC/DC", "Back in Black"}
print(a.union(b))                # {"AC/DC", "Back in Black", "thriller", "The Dark Side of the Moon"}
print(a.difference(b))            # {"Thriller"}
```

### Exercise 11 - A real question  ·  *hard*

Two product lists. Answer three business questions, each in one line.

```python
bought_tea    = {"Asha", "Ravi", "Meera", "Vikram"}
bought_coffee = {"Ravi", "Meera", "Sunil"}

print("Bought both:      ", bought_tea & bought_coffee)   # {"Ravi", "Meera"}
print("Tea only:         ", bought_tea - bought_coffee)   # {"Asha", "Vikram"}
print("Any customer:     ", bought_tea | bought_coffee)   # {"Asha", "Ravi", "Meera", "Vikram", "Sunil"}
print("Bought only one:  ", bought_tea ^ bought_coffee)   # {"Asha", "Vikram", "Sunil"}
```

### Exercise 12 - Word count  ·  *hard*

Count how many times each word appears in the sentence.
Use only what you know: `.split()`, `.lower()`, a dictionary and `.get()`.

```python
sentence = "the cat sat on the mat the cat slept"

# Using split() to split words in sentence
words = sentence.split()
print(f"List of Strings: {list(words)}")
# Creating a set to find unique elements
unique_words = set(words)
count_of_words = {w: words.count(w) for w in unique_words} # .count() will count words in the iterable

print(f"A set of coounts of words: {count_of_words}")
```
